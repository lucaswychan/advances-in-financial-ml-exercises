# Chapter 11 - The Dangers of Backtesting

## Preparation

## 1. An analyst fits an RF classifier where some of the features include seasonally adjusted employment data. He aligns with January data the seasonally adjusted value of January, etc. What “sin” has he committed?

It is the sin of look-ahead bias. Seasonal adjustment uses a two-sided filter — the algorithm (e.g., X-13ARIMA-SEATS) estimates the January factor using data from later months, so the seasonally adjusted January value embeds future information. Besides, the factors are revised annually, so the value we see today for a past January is not what was published then.

## 2. An analyst develops an ML algorithm where he generates a signal using closing prices, and executed at close. What's the sin?

It is a lookahead bias as well. We do not know the close before we execute it, so we cannot use it as a signal for the ML algorithm.

Besides, it ignores the transaction cost, where our cost must not be only the close, but also the maker fee (not the core sin tho).

## 3. There is a 98.51% correlation between total revenue generated by arcades and computer science doctorates awarded in the United States. As the number of doctorates is expected to grow, should we invest in arcades companies? If not, what's the sin?

Note that correlation doesn't imply causation. While they may have the same trend based on the historical data, it may not hold true for the future. The sin is storytelling.

## 4. The Wall Street Journal has reported that September is the only month of the year that has negative average stock returns, looking back 20, 50, and 100 years. Should we sell stocks at the end of August? If not, what's the sin?

We should not sell the stocks at the end of August. It introduces selection bias in multiple testing (data snooping). If we test all 12 months, one will appear "worst" by chance. Checking 20/50/100 years doesn't fix this because it's the same in-sample anomaly.W

Besides, the sin is storytelling as well, since the analysis is purely based on historical data without any theoretical and sound justification.

## 5. We download P/E ratios from Bloomberg, rank stocks every month, sell the top quartile, and buy the long quartile. Performance is amazing. What's the sin?

The first sin is the survivorship bias. Bloomberg's universe contains mostly companies that still exist, where dead/delisted firms are missing.

The second sin is look-ahead bias from non-PIT data. The downloaded P/E ratios reflect restated/backfilled earnings, not what was actually reported at the time we'd have traded.
